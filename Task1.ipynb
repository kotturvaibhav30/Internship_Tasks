```python
# Q30
# Ans: b is a separate list, so changing b does not change a.
a = [10, 20, 30]
b = a[:]
b[0] = 99
print(a)  # [10, 20, 30]
print(b)  # [99, 20, 30]
```

```python
# Q31
# Ans: Deleting the variable a does not delete the list still referenced by b.
a = [1, 2, 3]
b = a
del a
print(b)  # [1, 2, 3]
```

```python
# Q35
# Ans: A shallow copy shares nested lists, so changing b's inner list also changes a.
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)  # [[1, 99], [3, 4]]
```

```python
# Q36
# Ans: A deep copy duplicates nested lists, so changing b does not change a.
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)  # [[1, 2], [3, 4]]
```

```python
# Q37
# Ans: a.copy() is shallow, so b shares the inner lists that a modifies.
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)  # [[10], [20, 30]]
```

```python
# Q38
# Ans: deepcopy creates independent nested lists, so b remains unchanged.
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)  # [[10], [20]]
```

```python
# Q39
# Ans: The outer lists are different objects, but their first inner lists are shared.
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)       # False
print(a[0] is b[0]) # True
```

```python
# Q44
# Ans: *args collects positional arguments into a tuple.
def test(*args):
    print(args)

test(1, 2)  # (1, 2)
```

```python
# Q45
# Ans: args is a tuple because *args collects positional arguments into a tuple.
def test(*args):
    print(type(args))

test("a", "b", "c")  # <class 'tuple'>
```

```python
# Q46
# Ans: **kwargs collects keyword arguments into a dictionary.
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)  # {'name': 'Ram', 'age': 25}
```

```python
# Q47
# Ans: kwargs is a dictionary because **kwargs collects keyword arguments.
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)  # <class 'dict'>
```
